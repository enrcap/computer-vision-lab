### **Tarea B.1:** Defina y ejecute el método para cargar imágenes ``load_images()``

```python
def load_images(filenames: List) -> List:
    return [cv2.imread(filename) for filename in filenames]
```

```python
# TODO Build a list containing the paths of all images in the data folder
imgs_path = np.array([os.path.join("..", "data", f"00{i}.jpg") for i in range(6)])

imgs = load_images(imgs_path)
```

### **Tarea B.2**. Defina los métodos ``show_image()`` y ``write_image()`` para visualizar y guardar imágenes

```python
# TODO Complete the method, use every argument
def show_image(img: np.array, img_name: str = "Image"):
    cv2.imshow(img_name, img)
    cv2.waitKey()
    cv2.destroyAllWindows()

# TODO Complete the method, use every argument
def write_image(output_folder: str, img_name: str, img: np.array):
    os.makedirs(output_folder, exist_ok=True)
    img_path = os.path.abspath(os.path.join(output_folder, img_name))
    extension = os.path.splitext(img_name)[1] or ".jpg"
    success, encoded_img = cv2.imencode(extension, img)
    if not success:
        raise IOError(f"No se pudo codificar la imagen: {img_name}")
    encoded_img.tofile(img_path)
    return img_path
#esta funcion con chat porque OpenCV no guardaba bien en Windows porque mi ruta contiene caracteres como º y í.
#La escritura ahora usa cv2.imencode(...).tofile(...), compatible con rutas Unicode.
```

### **Tarea B.3:** Cambie el espacio de color de las imágenes a uno donde la crominancia y la intensidad estén separados (HSV)

```python
# TODO Get a list with ìmgs in HSV color space
hsv_imgs = [cv2.cvtColor(img, cv2.COLOR_RGB2HSV) for img in imgs]
```

### **Tarea B.4:** Segmente los colores anaranjados

```python
# TODO Define orange color range
light_orange = (1, 190, 200)
dark_orange = (255, 255, 255)

# TODO Compute a list that contains a mask (which should segment orange colors) for every image.
orange_masks = [cv2.inRange(img, light_orange, dark_orange) for img in hsv_imgs]

# TODO Compute a list that contains the result of multiplying the original image with its orange colors mask.
orange_segmented = [cv2.bitwise_and(img, img, mask=mask) for img, mask in zip(imgs, orange_masks)]

# TODO Show an original image
show_image(imgs[0])

# TODO Show a mask
show_image(orange_masks[0])

# TODO Show a segmented image
show_image(orange_segmented[0])
```

### **Tarea B.5:** Segmente los colores blancos.

Para detectar el rango de blancos complete la siguiente celda y ejecutela para investigar el rango de valores necesarios.

```python
# TODO Discover white color ranges
get_hsv_color_ranges(imgs[0])

# TODO Define white color range
light_white = (0, 0, 200)
dark_white = (179, 50, 255)

# TODO Compute a list that contains a mask (which should segment white colors) for every image.
white_masks = [cv2.inRange(img, light_white, dark_white) for img in hsv_imgs]

# TODO Compute a list that contains the result of multiplying the original image with its white colors mask.
white_segmented = [cv2.bitwise_and(img, img, mask=mask) for img, mask in zip(imgs, white_masks)]

# TODO Show an original image
show_image(imgs[0])

# TODO Show a mask
show_image(white_masks[0])

# TODO Show a segmented image
show_image(white_segmented[0])
```

### **Tarea B.6:** Junte las máscaras para cada imagen (naranja + blanco) y segmente cada una de ellas.

```python
# TODO Join orange_masks and white_masks
fish_masks =[cv2.bitwise_or(orange, white) for orange, white in zip(orange_masks, white_masks)]
    
# TODO Compute a list that contains the result of multiplying the original image with its complete mask.
fish = [cv2.bitwise_and(img, img, mask=mask) for img, mask in zip(imgs, fish_masks)]

# TODO Show an original image
show_image(imgs[0])

# TODO Show a mask
show_image(fish_masks[0])

# TODO Show a segmented image
show_image(fish[0])
```

### **Tarea B.7:** Guarde las imágenes

```python
# TODO Define your output folder and save every fish image
output_folder = os.path.abspath(os.path.join(os.getcwd(), "..", "output"))
os.makedirs(output_folder, exist_ok=True)
for i in range(len(imgs)):
    fish_img = cv2.bitwise_and(imgs[i], imgs[i], mask=fish_masks[i])
    write_image(output_folder, img_name=f"fish_00{i}.jpg", img=fish_img)
```

m### **Pregunta B.1:** Segmente por color el logotipo de su marca favorita: descompóngalo en al menos 3 colores.

```python
# Cargamos el escudo del Real Zaragoza en formato BGR
zaragoza_path = os.path.join("..", "data", "zaragoza_escudo.png")
zaragoza = cv2.imread(zaragoza_path)
if zaragoza is None:
    raise FileNotFoundError(f"No se pudo cargar la imagen: {zaragoza_path}")

zaragoza_hsv = cv2.cvtColor(zaragoza, cv2.COLOR_BGR2HSV)

# El fondo blanco queda fuera del total del escudo.
shield_mask = cv2.inRange(zaragoza_hsv, (0, 25, 20), (179, 255, 255))

# Definimos tres colores del escudo en HSV.
red_mask = cv2.bitwise_or(
    cv2.inRange(zaragoza_hsv, (0, 70, 50), (10, 255, 255)),
    cv2.inRange(zaragoza_hsv, (170, 70, 50), (179, 255, 255)),
)
yellow_mask = cv2.inRange(zaragoza_hsv, (15, 70, 70), (40, 255, 255))
blue_mask = cv2.inRange(zaragoza_hsv, (90, 60, 40), (135, 255, 255))

color_masks = {
    "Rojo": red_mask,
    "Amarillo": yellow_mask,
    "Azul": blue_mask,
}

# Aplicamos cada máscara sobre la imagen original.
segmented_colors = {
    color: cv2.bitwise_and(zaragoza, zaragoza, mask=mask)
    for color, mask in color_masks.items()
}

# Mostramos la imagen original, las máscaras binarias y los colores segmentados.
show_image(zaragoza, "Escudo del Real Zaragoza")
for color, mask in color_masks.items():
    show_image(mask, f"Mascara {color}")
    show_image(segmented_colors[color], f"{color} segmentado")

# Calculamos el porcentaje de cada color sobre los píxeles no blancos del escudo.
total_shield_pixels = cv2.countNonZero(shield_mask)
percentages = {
    color: 100 * cv2.countNonZero(mask) / total_shield_pixels
    for color, mask in color_masks.items()
}
classified_pixels = sum(cv2.countNonZero(mask) for mask in color_masks.values())
other_percentage = 100 * (total_shield_pixels - classified_pixels) / total_shield_pixels

print("Porcentaje de cada color sobre el total del escudo:")
for color, percentage in percentages.items():
    print(f"- {color}: {percentage:.2f}%")
print(f"- Otros colores/contornos: {other_percentage:.2f}%")
```
